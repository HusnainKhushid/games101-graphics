# A7 — multithread bonus

Each row on its own worker via `concurrent.futures`.

Sanity: 16spp Cornell went from 4m03 -> 34s on 8 cores.

```python
from concurrent.futures import ThreadPoolExecutor
import numpy as np

def render(scene, W, H, spp):
    img = np.zeros((H, W, 3))
    def do_row(j):
        row = np.zeros((W, 3))
        for i in range(W):
            for _ in range(spp):
                ray = scene.gen_ray(i, j, W, H)
                row[i] += cast_ray(scene, ray, 0)
            row[i] /= spp
        return j, row
    with ThreadPoolExecutor(max_workers=8) as ex:
        for j, row in ex.map(do_row, range(H)):
            img[j] = row
    return img
```
