## What this assignment is about

Warm-up. The interesting bit isn't the arithmetic — it's the trick of pushing 2D points into 3D as `(x, y, 1)` so that **translation becomes a matrix multiply**. Rotate, then translate, is one 3x3 matrix instead of a matrix + a vector add.

See: [Homogeneous coordinates — Wikipedia](https://en.wikipedia.org/wiki/Homogeneous_coordinates), [Rotation matrix — Wikipedia](https://en.wikipedia.org/wiki/Rotation_matrix).

### The task
Rotate P=(2,1) by 45° CCW around the origin, then translate by (1,2).


```python
import numpy as np

P = np.array([2.0, 1.0])
theta = np.deg2rad(45)
R = np.array([[np.cos(theta), np.sin(theta)],
              [-np.sin(theta), np.cos(theta)]])
# rotate then translate
Q = R @ P + np.array([1.0, 2.0])
print(Q)
```
