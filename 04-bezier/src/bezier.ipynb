## Topic

A **Bézier curve** is a smooth curve defined by a handful of control points. The
curve doesn't pass through the middle control points but is *pulled* toward them.
**De Casteljau's algorithm** evaluates the curve at parameter `t` by repeatedly
lerping neighboring points until one point remains — the point on the curve.

![bezier curve](https://upload.wikimedia.org/wikipedia/commons/d/d0/Bezier_curve.svg)
*A cubic Bézier and its control polygon (Wikipedia).*

See: [Bézier curve](https://en.wikipedia.org/wiki/B%C3%A9zier_curve), [De Casteljau's algorithm](https://en.wikipedia.org/wiki/De_Casteljau%27s_algorithm).

### Task
- `recursive_bezier(pts, t)` — de Casteljau.
- Sweep `t` in [0,1] and plot.
- **Bonus:** anti-alias the curve by weighting neighbouring pixels by distance.


```python
import numpy as np

def recursive_bezier(pts, t):
    if len(pts) == 1:
        return pts[0]
    new = [(1 - t) * a + t * b for a, b in zip(pts[:-1], pts[1:])]
    return recursive_bezier(new, t)

def bezier(pts, W, H):
    img = np.zeros((H, W))
    for t in np.linspace(0, 1, 1000):
        p = recursive_bezier(pts, t)
        x, y = int(round(p[0])), int(round(p[1]))
        if 0 <= x < W and 0 <= y < H:
            img[y, x] = 1.0
    return img
```
