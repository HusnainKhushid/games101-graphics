# A4 — Bezier + distance-weighted AA

```python
import numpy as np

def recursive_bezier(pts, t):
    if len(pts) == 1:
        return pts[0]
    new = [(1 - t) * a + t * b for a, b in zip(pts[:-1], pts[1:])]
    return recursive_bezier(new, t)

def bezier_aa(pts, W, H):
    img = np.zeros((H, W))
    for t in np.linspace(0, 1, 1000):
        p = recursive_bezier(pts, t)
        cx, cy = p[0], p[1]
        ix, iy = int(np.floor(cx)), int(np.floor(cy))
        # weight the 3x3 neighborhood by distance from the true point
        for dy in range(-1, 2):
            for dx in range(-1, 2):
                x, y = ix + dx, iy + dy
                if 0 <= x < W and 0 <= y < H:
                    d = np.hypot(cx - (x + 0.5), cy - (y + 0.5))
                    w = max(0.0, 1.0 - d / np.sqrt(2))
                    img[y, x] = max(img[y, x], w)
    return img
```
